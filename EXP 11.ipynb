```python
import matplotlib.pyplot as plt

marks = [45, 50, 55, 60, 65, 65, 70, 75, 80, 85]

plt.hist(marks, bins=5, edgecolor='black')
plt.title("Histogram of Students' Marks")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()
```

```python
import matplotlib.pyplot as plt

students = ['A', 'B', 'C', 'D']
marks = [70, 85, 60, 90]

plt.bar(students, marks)
plt.title("Students Marks")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.show()
```

```python
import matplotlib.pyplot as plt

subjects = ['Maths', 'Python', 'AI', 'DSA']
marks = [80, 90, 75, 85]

plt.pie(marks, labels=subjects, autopct='%1.1f%%')
plt.title("Subject Marks")
plt.show()
```

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 25]

plt.scatter(x, y)
plt.title("Scatter Plot")
plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.show()
```

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

plt.plot(x, y, marker='o')
plt.title("Simple Line Graph")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.grid(True)
plt.show()
```
