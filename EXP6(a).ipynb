```python
#set
empty_set=set()

print("Empty Set :",empty_set)

print("Type of empty_set:",type(empty_set))
```

```python
fruits={"apple","banana","cherry"}
print("set of fruits:",fruits)

mixed_set={1,"A",3.14,True}
print("Mixed set:", mixed_set)
```

```python
list_set=set([1,2,3,4])
print("Set from list:",list_set)

string_set=set("hello")
print("Set from string:",string_set)

tuple_set=set((10,20,30))
print("Set from Tuple :",tuple_set)
```

```python
#update()

fruits={"apple","banana","cherry"}
fruits.update(["mango","grapes"])
print("Set after update :",fruits)
```

```python
#add

fruits={"apple","banana","cherry"}
fruits.add("orange")
print("Set after adding :",fruits)
```

```python
#copy

original_set={"apple","banana","cherry"}
copied_set=original_set.copy()
print("Copied set:",copied_set)
```

```python
#pop

fruits={"apple","banana","cherry"}
popped_element=fruits.pop
print("Popped element:",popped_element)
print("set after pop:",fruits)

```

```python
#remove
fruits={"apple","banana","cherry"}
fruits.remove("banana")
print("Set after removing :",fruits)
```

```python
#discard

fruits={"apple","banana","cherry"}
fruits.discard("banana")
print("Set after discarding :",fruits)
```

```python
#clear

fruits={"apple","banana","cherry"}
fruits.clear()
print("Set after clear :",fruits)
```

```python
#union
#union using | operator

set1 ={"apple","banana","cherry"}
set2 ={"mango","grapes","banana"}

union_set_operator=set1 | set2
print("union using | operator:",union_set_operator)


#union using two sets

union_set=set1.union(set2)
print("union of set1 and set2:",union_set)
```

```python
#intersection
#intersection using two sets

set1 ={"apple","banana","cherry"}
set2 ={"mango","grapes","banana"}

intersection_set=set1.intersection(set2)
print("intersection of set1 and set2:",intersection_set)


#intersection using  & operator


intersection_set_operator=set1 & set2
print("intersection using & operator:",intersection_set_operator)

```

```python
#difference
#difference between two sets

set1 ={"apple","banana","cherry"}
set2 ={"mango","grapes","banana"}
difference_set=set1.difference(set2)
print("Difference of set1 and set2:",difference_set)

 # using - operator
    
difference_set_operator=set1 - set2
print("Difference using - operator:",difference_set_operator)



```
