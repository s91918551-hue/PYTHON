```python


#string operation

str1='dkte'
print(type(str1))

str2="dkte"
print(type(str2))

str3='''dkte'''
print(type(str3))

str4='10'
print(type(str4))
```

```python
#concatenation

str1='dkte' 
str2='college'

print(str1+str2)

print(str1+' '+str2)
```

```python

#len()

str1='dkte'
print(len(str1))

str2='dkte college'
print(len(str2))
```

```python

#indexing

str1='HELLO'
print(str1[0])
print(str1[1])
print(str1[2])
print(str1[3])
print(str1[4])
```

```python

#slicing

str1='dkte'
print(str1[1:3])
print(str1[0:3])
print(str1[:2])
print(str1[1:])
print(str1[:])
```

```python

str1='swalehin kausar amin shaikh'
print(str1[9:20])
print(str1[0:27])
print(str1[8:27])
```

```python

#negative slicing

str1='dkte'
print(str1[-1:-4])
```

```python

#string function

str1='Dkte'
print(str1.startswith('D'))

print(str1.endswith('e'))

print(str1.istitle())
```

```python

str1='dkte'
print(str1.isupper())

str2='DKTE'
print(str2.islower())
```
