```python
#if condition

age=20
if age>=18:
    print("Eligible to vote")
```

```python
#if....else statement

age=16
if age>=18:
    print("Eligible to vote")
else:
    print("Not Eligible to vote")
    
```

```python
#if...elif...else  statement

marks=85
if marks>=90:
    print("Grade A")
elif marks>=75:
    print("Grade B")
elif marks>=60:
    print("Grade C")
else:
    print("Grade D")
```

```python
#Nested if
age=25
citizen=True
if age>=18:
    if citizen:
        print("Eligible to vote")
```

```python
#logical operator
#and 
age=25
if age>=18 and age<=60:
    print("Working age")
    
```

```python
#or
marks=35
if marks>=30 or marks==35:
    print("Eligible for Grace Marks")
```
